## Questions de compréhension

### Question 1

Pourquoi ne donne-t-on pas directement le signal EEG brut à un classifieur classique comme LDA, SVM ou Random Forest ?

### Réponse 

## Pourquoi ne donne-t-on pas directement les signaux EEG bruts à un classifieur comme Random Forest ?

Les signaux EEG bruts ne sont généralement pas utilisés directement comme entrée d’un classifieur tel qu’un **Random Forest** car ils présentent plusieurs difficultés :

### 1. Très grande dimension des données

Les signaux EEG sont échantillonnés à une fréquence élevée et contiennent un grand nombre d’échantillons temporels. Une fenêtre de quelques secondes peut ainsi contenir des centaines ou des milliers de valeurs par canal.

Le nombre de variables devient alors très important par rapport au nombre d’exemples disponibles, ce qui complique l’apprentissage et augmente le risque de surapprentissage (*overfitting*).

### 2. Présence importante de bruit

Les enregistrements EEG sont sensibles à de nombreux artefacts :

- mouvements de la tête ;
- clignements des yeux ;
- activité musculaire ;
- perturbations électriques de l’environnement.

Les valeurs brutes contiennent donc beaucoup d’informations non pertinentes pour la tâche de classification.

### 3. Information peu exploitable directement

La charge cognitive est davantage liée à certaines caractéristiques du signal (par exemple la puissance dans les bandes de fréquences **delta**, **theta**, **alpha**, **beta** ou **gamma**) qu’aux valeurs instantanées du signal EEG.

Les échantillons bruts ne mettent pas directement en évidence ces propriétés physiologiques.

### 4. Limitation des modèles classiques

Les algorithmes comme **Random Forest**, **SVM** ou **XGBoost** fonctionnent mieux avec un ensemble réduit de variables descriptives (*features*) qu’avec des séries temporelles longues et fortement corrélées.

On extrait donc généralement des caractéristiques telles que :

- la puissance spectrale par bande de fréquence ;
- les statistiques temporelles (moyenne, variance, énergie, entropie, etc.) ;
- des indicateurs de connectivité ou de complexité du signal.

### Conclusion

Avant d'entraîner un classifieur classique comme **Random Forest**, les signaux EEG sont généralement prétraités puis transformés en caractéristiques représentatives. Cette étape réduit le bruit, diminue la dimension des données et met en évidence les informations les plus pertinentes pour l’estimation de la charge cognitive.

### Question 2

Pourquoi les features fréquentielles sont-elles particulièrement importantes en EEG ?

### Réponse 

## Pourquoi les features fréquentielles sont-elles particulièrement importantes en EEG ?

Les **features fréquentielles** sont particulièrement importantes en EEG car une grande partie de l’information physiologique pertinente est portée par les **oscillations cérébrales** présentes dans différentes bandes de fréquences.

En effet, l’EEG mesure l’activité électrique du cerveau sous forme de signaux qui oscillent à différentes fréquences. Ces fréquences sont associées à différents états cognitifs et physiologiques.

### Principales bandes de fréquences EEG

- **Delta (0,5–4 Hz)** : sommeil profond et états de faible vigilance.
- **Theta (4–8 Hz)** : mémoire de travail, attention et effort mental.
- **Alpha (8–13 Hz)** : relaxation et diminution de l’attention visuelle.
- **Beta (13–30 Hz)** : concentration, activité cognitive et vigilance.
- **Gamma (> 30 Hz)** : traitement complexe de l’information et intégration cognitive.

### Lien avec la charge cognitive

Lorsque la charge cognitive varie, la répartition de l’énergie entre ces bandes de fréquences change également.

Par exemple :

- une augmentation de la charge cognitive est souvent associée à une augmentation de l’activité **theta frontale** ;
- une diminution de la puissance **alpha** est fréquemment observée lorsque l’attention et l’effort mental augmentent ;
- certaines variations de l’activité **beta** peuvent également refléter un engagement cognitif plus important.

Ces modifications sont souvent plus informatives que les amplitudes brutes du signal EEG.

### Pourquoi utiliser des features fréquentielles ?

L’analyse fréquentielle permet de résumer le signal EEG sous forme d’indicateurs pertinents tels que :

- la puissance spectrale dans chaque bande de fréquence ;
- la puissance relative entre bandes ;
- les rapports entre bandes (par exemple theta/alpha ou theta/beta).

Ces caractéristiques :

- réduisent la dimension des données ;
- sont plus robustes au bruit ;
- possèdent une interprétation neurophysiologique claire ;
- améliorent généralement les performances des modèles de classification.

### Conclusion

Les features fréquentielles sont essentielles en EEG car elles capturent directement les rythmes cérébraux associés aux processus cognitifs. Pour l’estimation de la charge cognitive, elles fournissent souvent des informations plus discriminantes et plus interprétables que les signaux EEG bruts.

## 5. Entropie spectrale

### Principe théorique

L'entropie spectrale mesure la dispersion de l'énergie dans le domaine fréquentiel.

On calcule d'abord la PSD dans une bande, puis on la normalise pour obtenir une distribution :

$$
p_i = \frac{PSD_i}{\sum_j PSD_j}
$$

L'entropie de Shannon est ensuite :

$$
H = - \sum_i p_i \log(p_i)
$$


### Fonction Python utile

```python
from scipy.stats import entropy
```

`entropy(p)` calcule directement l'entropie de Shannon si `p` est une distribution normalisée.

## 6. Paramètres de Hjorth : mobility et complexity

Les paramètres de Hjorth sont des descripteurs temporels utilisés pour caractériser la dynamique d'un signal.

### Mobility

La mobility mesure approximativement la fréquence moyenne du signal :

$$
Mobility(x) = \sqrt{\frac{Var(\Delta x)}{Var(x)}}
$$

où $\Delta x$ est la dérivée discrète du signal, généralement calculée avec `np.diff(x)`.

### Complexity

La complexity mesure la variation de la mobility entre le signal et sa dérivée :

$$
Complexity(x) = \frac{Mobility(\Delta x)}{Mobility(x)}
$$

## 7. Complexité de Lempel-Ziv

### Principe théorique

La complexité de Lempel-Ziv mesure le nombre de motifs nouveaux rencontrés dans une séquence.

Comme l'algorithme s'applique à une séquence symbolique, un signal EEG réel doit d'abord être transformé en séquence binaire.

Une méthode simple consiste à binariser le signal par rapport à sa médiane :

$$
b[n] =
\begin{cases}
1, & x[n] > median(x) \\
0, & x[n] \le median(x)
\end{cases}
$$


### Paramètres importants

- seuil de binarisation : médiane ou moyenne ;
- normalisation de la complexité pour comparer des segments de même ou de différente longueur ;

## 8. Dimension fractale de Higuchi

### Principe théorique

La dimension fractale de Higuchi cherche à mesurer la complexité géométrique d'un signal temporel.

Un signal très lisse ressemble davantage à une courbe régulière. Un signal très irrégulier ou bruité présente une trajectoire plus complexe.

L'algorithme de Higuchi construit plusieurs sous-séquences avec différents pas $k$, mesure leur longueur moyenne $L(k)$, puis estime une pente dans un espace logarithmique.

### Paramètre important

- `kmax` : pas maximal testé.

Pour un segment de 10 secondes à 256 Hz, une valeur pédagogique simple est :

```python
kmax = 10
```

Une valeur trop faible peut donner une estimation instable ; une valeur trop élevée augmente le coût de calcul.
