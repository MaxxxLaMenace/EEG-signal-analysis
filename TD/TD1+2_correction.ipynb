## 1. Contexte général

La conduite automobile est une activité complexe qui mobilise plusieurs ressources cognitives : perception visuelle, attention soutenue, prise de décision, anticipation, contrôle moteur et gestion des imprévus.

Le papier CL-Drive s'intéresse donc à une question centrale :

> Peut-on estimer automatiquement la charge cognitive d'un conducteur à partir de signaux physiologiques et comportementaux mesurés pendant la conduite ?

#### Q1. Pourquoi est-il intéressant d'estimer la charge cognitive d'un conducteur ?

**Réponse :**

Si la charge cognitive augmente au-delà d'un certain point, les performances du conducteur se dégradent, ce qui peut augmenter la probabilité d'accidents de la route. L'estimation de cette charge permettrait aux technologies intelligentes intégrées aux véhicules de déclencher une alarme si des niveaux dangereusement élevés sont détectés, réduisant ainsi le nombre de collisions.

## 2. Définition : charge cognitive

#### Q2. Comment définir la charge cognitive ?

**Réponse :**

La charge cognitive désigne la quantité d'informations que notre mémoire de travail peut traiter à un instant donné. Autrement dit, c'est la quantité de ressources cognitives nécessaires pour accomplir une tâche.

#### Q3. Dans le contexte de la conduite, de quels facteurs la charge cognitive peut-elle dépendre ?

**Réponse :**

Elle dépend de facteurs intrinsèques liés à la complexité inhérente de la conduite, de facteurs extrinsèques environnementaux qui ne sont pas pertinents pour la tâche, ainsi que de l'engagement dans des tâches secondaires comme l'utilisation d'un téléphone mobile ou une conversation verbale.

Dans le papier, la charge cognitive est évaluée à l'aide d'une échelle subjective appelée **PAAS**, allant de 1 à 9.

## 3. Échelle PAAS utilisée pour les labels

Les participants rapportent leur charge cognitive sur une échelle de 1 à 9.

**Réponse :**

Il s'agit d'une vérité terrain subjective car elle repose sur l'auto-évaluation (self-reporting) des participants. Ce sont les conducteurs eux-mêmes qui déclarent verbalement la charge qu'ils ressentent.

## 5. Modalités mesurées

Le dataset CL-Drive est **multimodal**. Il contient plusieurs types de signaux enregistrés pendant la même tâche de conduite.

#### Q7. Compléter le tableau suivant.

| Modalité | Signification | Signal mesuré | Rôle possible |
|---|---|---|---|
| EEG | Électroencéphalogramme | Activité électrique du cerveau | Capter les changements d'activité cérébrale lors des variations de charge cognitive |
| ECG | Électrocardiographie | Rythme cardiaque | Mesurer la variabilité de la fréquence cardiaque contrôlée par le système nerveux autonome |
| EDA | Activité Électrodermale | Conductance de la peau (transpiration) | Mesurer la réponse contrôlée par le système nerveux sympathique face à l'effort mental |
| Gaze | Suivi du regard | Mouvements oculaires, clignements, pupilles | Observer les corrélations entre la taille de la pupille ou les fixations et la charge cognitive |

Dans le projet, la modalité principale sera l’**EEG**. Les autres modalités serviront éventuellement d’extensions pour discuter de la fusion multimodale.

#### Q8. Pourquoi peut-il être intéressant de combiner EEG, ECG, EDA et Gaze pour estimer la charge cognitive ?

**Réponse :**

Les approches multimodales ont prouvé leur efficacité pour augmenter la précision de l'évaluation de la charge cognitive. Dans cette étude, la combinaison simultanée des quatre capteurs permet à l'IA d'obtenir les meilleurs résultats de classification par rapport aux approches utilisant un seul capteur.

## 7. Protocole expérimental

#### Q11. Quelle est la différence entre un **protocole expérimental** et un **dataset exploitable en machine learning** ?

**Réponse :**

Le protocole expérimental décrit les étapes physiques de l'expérience et d'acquisition des données avec les participants (temps de repos, ordre des tâches, conditions). Le dataset correspond aux données une fois récoltées, nettoyées, formatées et associées à leurs labels, prêtes à nourrir les algorithmes de machine learning.

#### Q12. Donner une description détaillée du protocole expérimental utilisé.

**Réponse :**

L'expérience débute par 3 minutes de collecte d'une ligne de base (baseline), suivies d'un scénario d'orientation. Ensuite, pour chaque scénario de conduite, le cycle est le suivant : 2 minutes de repos, puis 2 minutes de nouvelle baseline, et enfin le scénario de conduite de 3 minutes. Pendant ce scénario, une alarme sonore invite le participant à évaluer sa charge cognitive toutes les 10 secondes.

#### Q13. Pourquoi les auteurs collectent-ils une baseline avant les scénarios ?

**Réponse :**

Les données de la ligne de base (baseline) sont collectées pour être utilisées ultérieurement afin de normaliser les signaux physiologiques de chaque individu.

#### Q14. Pourquoi prévoir un scénario d’orientation avant les scénarios principaux ?

**Réponse :**

Le scénario d'orientation permet aux participants de s'adapter au simulateur, de bien comprendre le système d'évaluation des charges, et surtout de minimiser le risque d'apparition du Syndrome d'Adaptation au Simulateur (SAS).

#### Q15. Pourquoi prévoir des périodes de repos entre les scénarios ?

### Réponse

Pour permettre aux participants de se reposer, de réduire la possibilité de ressentir le Syndrome d'Adaptation au Simulateur (SAS), et de revenir à un état physiologique et cognitif neutre avant la tâche suivante.

#### Q16. Pourquoi demander un score de charge cognitive toutes les 10 secondes plutôt qu’une seule fois à la fin du scénario ?

### Réponse

Des évaluations denses et fréquentes permettent de capter l'évolution dynamique de la charge mentale en cours de tâche, générant plus de données fiables pour le modèle ML. Attendre la fin de scénarios de 3 minutes pourrait amener les participants à oublier ce qu'ils ont ressenti au début.

## 8. Scénarios de conduite

Le dataset contient 9 scénarios principaux, conçus pour induire différents niveaux de charge cognitive.

#### Q17. Compléter le tableau suivant.

| Scénario | Description synthétique |
|---:|---|
| 1 | Conduite sur autoroute |
| 2 | Conduite de nuit |
| 3 | Conduite de nuit sur autoroute avec de la neige |
| 4 | Défi de la balle de tennis (toucher la balle avec le pneu) |
| 5 | Défi du slalom (naviguer entre des portes) |
| 6 | Défi du passage étroit |
| 7 | Défi des virages à 90 degrés à gauche et à droite |
| 8 | Défi du demi-tour en 3 points |
| 9 | Défi de navigation dans une ruelle étroite |

L'hypothèse générale est que les scénarios les plus complexes induisent une charge cognitive plus élevée.

#### Q18. Pourquoi utiliser plusieurs scénarios de conduite au lieu d'un seul scénario ?

**Réponse :**

Afin de solliciter le conducteur de différentes manières et d'induire une grande variété de niveaux de charge cognitive pour constituer un jeu de données diversifié.

#### Q19. Pourquoi peut-on s’attendre à une charge cognitive plus élevée dans les scénarios 8 et 9 ?

### Réponse

Ces scénarios sont conçus pour être les plus complexes mécaniquement et visuellement. La carte thermique (heat map) de l'étude confirme que les participants rapportent très majoritairement des scores élevés lors de ces derniers défis.

#### Q20. La complexité du scénario suffit-elle à définir le label de charge cognitive ?

### Réponse

Non, car la charge cognitive dépend également des ressources individuelles, du niveau de stress et des compétences propres à chaque conducteur. La difficulté induite par la tâche s'additionne aux facteurs environnementaux, c'est pourquoi un ressenti subjectif (PAAS) est requis.

## 9. Échantillonnage et segmentation temporelle

Les signaux EEG sont continus. Pour entraîner un modèle, il faut transformer ce signal continu en exemples individuels.

Dans CL-Drive, les scores PAAS de charge cognitive sont collectés toutes les 10 secondes.

Il est donc naturel de découper les signaux physiologiques en fenêtres de 10 secondes :

$$
X_i = \text{segment de 10 s}
$$

$$
y_i = \text{score PAAS associé}
$$

On obtient ainsi le problème suivant :

Le principe d’association segment-label est le suivant :

| Temps | Segment EEG | Score PAAS |
|---|---|---|
| 0–10 s | segment 1 | score 1 |
| 10–20 s | segment 2 | score 2 |
| 20–30 s | segment 3 | score 3 |

#### Q21. Quelle est la fréquence d’échantillonnage EEG utilisée dans le papier ? 


**Réponse :**

256 Hz.

#### Q22. Combien d’échantillons contient une fenêtre EEG de 10 secondes ?


**Réponse :**

Sachant que la fréquence est de 256 Hz (256 échantillons par seconde), une fenêtre de 10 secondes contient 2560 échantillons par canal.

#### Q23. Combien de valeurs contient une fenêtre EEG de 10 secondes si les 4 canaux sont utilisés ?

**Réponse :**

2560 échantillons multipliés par 4 canaux donnent un total de 10 240 valeurs brutes pour une seule fenêtre de 10 secondes.

#### Q24. Un scénario dure 3 minutes. Combien de fenêtres de 10 secondes peut-on obtenir par scénario ?

### Réponse

3 minutes équivalent à 180 secondes. On obtient donc 18 segments de 10 secondes par scénario.

#### Q25. Si un sujet réalise 9 scénarios complets, combien de fenêtres obtient-on pour ce sujet ?

### Réponse

18 fenêtres multipliées par 9 scénarios = 162 fenêtres par sujet.

#### Q26. Si 21 sujets réalisent tous 9 scénarios complets, combien de fenêtres obtient-on au total ?

### Réponse

162 fenêtres multipliées par 21 sujets = 3402 fenêtres au total.

## 10. Formulation du problème d'apprentissage automatique

À partir du dataset CL-Drive, on peut formuler plusieurs problèmes de classification.

### Classification binaire

| Scores PAAS | Classe |
|---|---|
| 1 à 4 | charge faible |
| 5 à 9 | charge élevée |

### Classification ternaire

| Scores PAAS | Classe |
|---|---|
| 1 à 3 | charge faible |
| 4 à 6 | charge moyenne |
| 7 à 9 | charge élevée |

Dans le projet, on commencera par la classification binaire.

#### Q27. Pourquoi commencer par une classification binaire plutôt qu’une classification ternaire ?

### Réponse

Les auteurs constatent que la classification ternaire est plus difficile à prédire que la binaire. De plus, simplifier le problème en deux classes (faible vs élevée) est souvent la meilleure première étape lors de l'entraînement d'un nouveau modèle.

#### Q28. Dans une classification binaire, pourquoi le score PAAS 5 est-il généralement placé dans la classe « charge élevée » dans ce papier ?

### Réponse

C'est un choix méthodologique arbitraire des auteurs : regrouper les notes de 1 à 4 dans la catégorie "faible" et les notes de 5 à 9 dans la catégorie "élevée" permet d'équilibrer les classes ou de placer un seuil clair de vigilance médiane.

#### Q29. Quel est le lien entre ce papier et un système d'interface adaptative ?

**Réponse :**

L'objectif technologique final de ces recherches est de développer des systèmes d'interface pour les véhicules intelligents. Une interface adaptative pourrait surveiller en temps réel la charge cognitive du conducteur (via les capteurs) et adapter le comportement de la voiture ou déclencher une alarme si un seuil de surcharge mentale est franchi.

## 9. Visualisation des signaux EEG

Avant tout prétraitement ou apprentissage automatique, il est indispensable de visualiser les signaux EEG.

La visualisation permet de :

- vérifier que les données sont correctement chargées ;
- observer les amplitudes ;
- repérer les valeurs aberrantes ;
- détecter les ruptures ou données manquantes ;
- comparer les canaux ;
- repérer des artefacts évidents ;
- comparer qualitativement des segments de faible et forte charge cognitive.

### Algorithme conceptuel pour visualiser un segment EEG

Pour visualiser un segment EEG de 10 secondes :

1. choisir un sujet ;
2. choisir un scénario ;
3. choisir un segment ;
4. récupérer les 4 canaux EEG ;
5. calculer le nombre d’échantillons : $N = f_s \times T$ ;
6. construire l’axe temporel : $t[n] = n/f_s$ ;
7. tracer les quatre canaux en fonction du temps ;
8. ajouter le score PAAS associé au titre du graphique ;
9. commenter la qualité du signal.

#### Q30. Que faut-il regarder en premier lorsqu’on affiche un signal EEG brut ?

### Réponse

Il faut observer l'amplitude globale pour repérer des anomalies ou pertes de connexion évidentes, ainsi que le bruit caractéristique (comme l'interférence électrique) nécessitant un filtrage ultérieur.

#### Q31. Pourquoi afficher les quatre canaux EEG plutôt qu’un seul ?

### Réponse

Les artefacts (comme un mouvement des yeux ou un clignement) n'affectent pas toujours les électrodes frontales de la même manière que les temporales. Comparer les quatre signaux permet de différencier un bruit très localisé d'un véritable signal cérébral global.

#### Q32. Pourquoi comparer visuellement un segment de faible charge et un segment de forte charge cognitive ?

### Réponse

Afin de s'assurer intuitivement qu'il y a de réelles différences comportementales dans les ondes (par exemple en termes d'amplitude ou de fréquences dominantes) avant même de déléguer la classification au machine learning.

#### Q33. Pourquoi l'utilisation d'un simulateur est-elle à la fois un avantage et une limite ?

**Réponse :**

* **Avantage :** L'usage du simulateur offre une alternative sans risque physique réel, en plus de permettre la création de scénarios contrôlés et reproductibles à volonté (gestion de la météo, du trafic).
* **Limite :** Certains participants ressentent le Syndrome d'Adaptation au Simulateur (SAS), ce qui a provoqué des abandons et la nature "virtuelle" ne capture pas 100% de la tension d'une véritable conduite.

## Activité 3 — Mini-rendu TD1

Rédiger un court résumé du papier en 10 à 15 lignes.

Le document introduit "CL-Drive", une base de données multimodale conçue pour analyser la charge cognitive des conducteurs automobiles. Le problème soulevé est le risque de baisse de performance en cas de surcharge mentale au volant. L'expérience s'appuie sur l'utilisation d'un simulateur immersif reproduisant 9 scénarios de complexité variable (allant de l'autoroute dégagée à la conduite urbaine dans une ruelle étroite). 21 participants ont été équipés de capteurs enregistrant quatre signaux physiologiques : l'activité cérébrale (EEG), cardiaque (ECG), électrodermale (EDA) et visuelle (Gaze).

Les chercheurs démontrent qu'en utilisant des modèles de Machine Learning de classification (pour classer les données en binaire: charge faible ou élevée), l'algorithme performe avec une grande précision, particulièrement en croisant toutes les modalités. Cet ensemble de données est une base fondamentale dans la conception de futures interfaces véhicules adaptatives qui sauront détecter la surcharge mentale en temps réel et alerter le conducteur.
