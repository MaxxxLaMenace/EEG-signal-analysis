## 1. Contexte général

La conduite automobile est une activité complexe qui mobilise plusieurs ressources cognitives : perception visuelle, attention soutenue, prise de décision, anticipation, contrôle moteur et gestion des imprévus.

Le papier CL-Drive s'intéresse donc à une question centrale :

> Peut-on estimer automatiquement la charge cognitive d'un conducteur à partir de signaux physiologiques et comportementaux mesurés pendant la conduite ?

#### Q1. Pourquoi est-il intéressant d'estimer la charge cognitive d'un conducteur ?

**Réponse :**

Estimer la charge cognitive d’un conducteur est intéressant car cela permet d’évaluer en temps réel son niveau de sollicitation mentale et donc d’améliorer la sécurité routière.

Dans l’article Multimodal Brain-Computer Interface for In-Vehicle Driver Cognitive Load Measurement: Dataset and Baselines, les auteurs expliquent que certaines situations de conduite ou certaines tâches secondaires (navigation, téléphone, systèmes embarqués, multitâche, etc.) augmentent fortement la charge cognitive. Lorsque cette charge devient trop élevée, les capacités du conducteur diminuent : attention réduite, temps de réaction plus long, moins bonne prise de décision et augmentation du risque d’accident.

L’estimation de cette charge cognitive permet alors :

d’adapter les systèmes embarqués du véhicule ;
de limiter les distractions lorsque le conducteur est déjà très sollicité ;
d’améliorer les systèmes d’assistance à la conduite ;
de détecter des situations dangereuses avant qu’une erreur humaine ne se produise ;
et, dans les véhicules semi-autonomes, de vérifier si le conducteur est capable de reprendre le contrôle du véhicule rapidement et en sécurité.

Le document souligne aussi qu’une mesure fiable de la charge cognitive peut être obtenue grâce à plusieurs signaux physiologiques combinés (EEG, ECG, activité électrodermale, suivi des yeux), ce qui ouvre la voie à des véhicules capables d’adapter leur comportement à l’état mental du conducteur.

## 2. Définition : charge cognitive

#### Q2. Comment définir la charge cognitive ?

**Réponse :**

La charge cognitive correspond à la quantité de ressources mentales qu’une personne doit mobiliser pour effectuer une tâche.

Dans le contexte de la conduite automobile, elle représente l’effort mental nécessaire pour :

percevoir l’environnement ;
traiter les informations ;
prendre des décisions ;
et réaliser les actions de conduite.

Plus une situation est complexe (trafic dense, météo difficile, multitâche, utilisation d’un GPS ou d’un téléphone, etc.), plus la charge cognitive augmente.

D’après les travaux présentés dans l’article Multimodal Brain-Computer Interface for In-Vehicle Driver Cognitive Load Measurement: Dataset and Baselines, la charge cognitive est liée à la capacité limitée du cerveau humain à traiter simultanément plusieurs informations. Lorsqu’elle devient trop élevée, les performances diminuent : baisse de l’attention, augmentation du temps de réaction et risque accru d’erreurs.

On peut donc définir la charge cognitive comme :

le niveau de sollicitation des ressources mentales d’un individu lors de l’exécution d’une tâche.

En conduite, cette notion est particulièrement importante car un conducteur surchargé cognitivement peut perdre en vigilance ou ne plus réagir correctement à une situation dangereuse.

#### Q3. Dans le contexte de la conduite, de quels facteurs la charge cognitive peut-elle dépendre ?

**Réponse :**

#### Facteurs influençant la charge cognitive en conduite

Dans le contexte de la conduite, la charge cognitive dépend de nombreux facteurs liés à la fois au conducteur, à la tâche de conduite et à l’environnement.

L’article *Multimodal Brain-Computer Interface for In-Vehicle Driver Cognitive Load Measurement: Dataset and Baselines* met notamment en évidence que la charge cognitive augmente lorsque le conducteur doit traiter davantage d’informations ou réaliser plusieurs tâches simultanément.

Parmi les principaux facteurs :

- **La complexité de la situation de conduite**  
  Circulation dense, intersections, dépassements, conduite de nuit ou mauvaises conditions météorologiques demandent davantage d’attention et de prise de décision.

- **Les tâches secondaires**  
  Utiliser un GPS, répondre à un appel, interagir avec l’écran du véhicule ou discuter avec des passagers augmente fortement la sollicitation mentale.

- **Le niveau de distraction**  
  Les distractions visuelles, auditives ou cognitives détournent une partie des ressources mentales nécessaires à la conduite.

- **L’état du conducteur**  
  La fatigue, le stress, le manque de sommeil, les émotions ou encore l’expérience de conduite influencent directement la capacité cognitive disponible.

- **La durée de conduite**  
  Une longue période de conduite peut entraîner une fatigue mentale et modifier la charge cognitive au fil du temps.

- **Le niveau d’automatisation du véhicule**  
  Dans les véhicules semi-autonomes, la surveillance du système et la nécessité éventuelle de reprendre le contrôle peuvent générer une charge cognitive particulière.

- **La quantité d’informations à traiter**  
  Les alertes du véhicule, les panneaux, les autres usagers et les événements imprévus augmentent le volume d’informations que le conducteur doit analyser rapidement.

Le document souligne également que ces facteurs peuvent être étudiés à l’aide de signaux physiologiques (EEG, ECG, suivi oculaire, activité électrodermale), car ils reflètent les variations de l’effort mental du conducteur.

Dans le papier, la charge cognitive est évaluée à l'aide d'une échelle subjective appelée **PAAS**, allant de 1 à 9.

#### Q5. Pourquoi parle-t-on ici d'une vérité terrain subjective ?

**Réponse :**

#### Pourquoi parle-t-on d’une vérité terrain subjective ?

Dans ce contexte, on parle de **vérité terrain subjective** (*subjective ground truth*) car la charge cognitive n’est pas une grandeur directement observable ou mesurable de manière absolue.

Contrairement à une mesure physique comme la vitesse ou la température, la charge cognitive correspond à un état mental interne du conducteur. Il est donc difficile d’obtenir une valeur « exacte » et universelle de cette charge.

Dans les études comme celle présentée dans *Multimodal Brain-Computer Interface for In-Vehicle Driver Cognitive Load Measurement: Dataset and Baselines*, la vérité terrain est souvent construite à partir :

- de questionnaires d’auto-évaluation ;
- du ressenti du conducteur ;
- d’échelles psychologiques comme la NASA-TLX ;
- ou de l’association entre certaines tâches et un niveau supposé de difficulté cognitive.

Ainsi, deux conducteurs placés dans une même situation peuvent ressentir une charge cognitive différente selon :
- leur expérience ;
- leur fatigue ;
- leur stress ;
- ou leurs capacités individuelles.

La « vérité » utilisée pour entraîner ou évaluer les modèles dépend donc en partie de la perception humaine et de conventions expérimentales, ce qui la rend subjective.

Cette subjectivité constitue un enjeu important pour les systèmes d’estimation de charge cognitive, car les signaux physiologiques (EEG, ECG, suivi oculaire, etc.) doivent être interprétés à partir de labels qui ne sont pas parfaitement objectifs.

## 9. Échantillonnage et segmentation temporelle

Dans CL-Drive, les scores PAAS de charge cognitive sont collectés toutes les 10 secondes.

Il est donc naturel de découper les signaux physiologiques en fenêtres de 10 secondes :

$$
X_i = \text{segment de 10 s}
$$

$$
y_i = \text{score PAAS associé}
$$

## 9. Visualisation des signaux EEG

Avant tout prétraitement ou apprentissage automatique, il est indispensable de visualiser les signaux EEG.

La visualisation permet de :

- vérifier que les données sont correctement chargées ;
- observer les amplitudes ;
- repérer les valeurs aberrantes ;
- détecter les ruptures ou données manquantes ;
- comparer les canaux ;
- repérer des artefacts évidents ;
- comparer qualitativement des segments de faible et forte charge cognitive.

### Algorithme conceptuel pour visualiser un segment EEG

Pour visualiser un segment EEG de 10 secondes :

1. choisir un sujet ;
2. choisir un scénario ;
3. choisir un segment ;
4. récupérer les 4 canaux EEG ;
5. calculer le nombre d’échantillons : $N = f_s \times T$ ;
6. construire l’axe temporel : $t[n] = n/f_s$ ;
7. tracer les quatre canaux en fonction du temps ;
8. ajouter le score PAAS associé au titre du graphique ;
9. commenter la qualité du signal.
